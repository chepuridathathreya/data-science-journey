```python
import matplotlib.pyplot as plt

time = [8, 7, 9, 5, 8, 7, 10, 11, 12, 14, 4, 3]

plt.figure(figsize=(10, 5))
plt.hist(time, bins=4)

plt.xlabel("Time spent")
plt.ylabel("No. of students")
plt.title("Time spent by students on mobile")
plt.grid()

plt.show()
```
