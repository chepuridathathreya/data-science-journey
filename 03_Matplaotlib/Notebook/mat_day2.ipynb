```python
import matplotlib.pyplot as plt
```

```python
years = [2001,2002,2003,2004,2005]
birthrate = [12,14,29,12,21]
deathrate = [11,10,2,13,23]
plt.plot(years, birthrate, label = "Birth rate")
plt.plot(years, deathrate, label = "Death rate")
plt.xlabel("Years")
plt.ylabel("Rate")
plt.legend()
plt.show()
plt.xticks(years)
```

```python
months = ["jan","feb","mar","apr","may"]
study_hrs = [100,95,98,92,89]
exersice_hrs = [40,112,128,89,115]
plt.plot(months, study_hrs, label = "study")
plt.plot(months, exersice_hrs, label = "workout")
plt.xlabel("Months")
plt.ylabel("Time spent")
plt.legend()
plt.show
```

```python
subjects = ["python", "sql","pandas", "ML"]
hrs = [90, 29, 12, 130]
plt.bar(subjects, hrs)
plt.xlabel("subjects")
plt.ylabel("time spent")
plt.title("Learning hrs")
plt.show()
```

```python
anime = ["one piece", "Fighting spirit", "JJK", "jojo", "Demon slayer"]
sale = [2099, 1100, 1000, 1872, 1223]
plt.bar(anime, sale)
plt.xlabel("Anime name")
plt.ylabel("Manga sold")
plt.title("Sales of manga in a month")
```

```python
insta = [2, 4, 6, 8]
sleep = [8, 6, 4, 2]
plt.scatter(insta, sleep)
plt.xlabel("Insta screen time")
plt.ylabel("Sleep in hrs")
plt.title("effect of sleep per screen time CASE: INSTA")
plt.show()
```

```python
marks = [45, 52, 61, 68, 72, 75, 78, 81, 83, 85,
         87, 88, 89, 91, 92, 94, 95, 96, 98, 99]

plt.hist(marks, bins=5)

plt.xlabel("Marks")
plt.ylabel("Number of students")
plt.title("Distribution of Student Marks")
```

```python
#TIME SPENT BY INDEIVIDUALS A DAY ON MBL
time = [8,7,9,5,8,7,10,11,12,14,4,3]
plt.hist(time, bins = 4)
plt.figure(figsize=(10, 20))
plt.xlabel("Time spent")
plt.ylabel("No.of studuents ")
plt.title("Timespent by students on mbl")
plt.grid()
plt.show()
```
