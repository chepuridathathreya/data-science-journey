```python
days = [1,2,3,4,5,6,7]
water = [2,3,5,1,5,6,2]
plt.plot(days,water)
```

```python
plt.xlabel("Days")
plt.ylabel("Water intake")
plt.title("liter's of water per day")
plt.plot(days,water)
plt.show()
```

```python
days = [1,2,3,4,5,6,7]
hours = [2,4,3,6,1,2,3]
```

```python
plt.plot(days, hours, label="Study Hours")
plt.xlabel("Day")
plt.ylabel("Hours")
plt.title("Weekly Study")
plt.legend()
```

```python
years = [2001,2002,2003,2004,2005]
birthrate = [20,10,30,40,10]
plt.plot(years, birthrate, label ="birth rates per years in thousands" )
plt.xlabel("Year")
plt.ylabel("Birth rate")
plt.legend()
```
