```python
import matplotlib.pyplot as plt
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

study_hours = [40, 45, 52, 48, 60, 65]
workout_hours = [12, 15, 18, 14, 20, 22]
sleep_hours = [7.5, 7, 6.5, 7.2, 7.8, 8]

marks = [55, 62, 68, 71, 78, 85, 88, 91, 73, 64, 82, 95]

screen_time = [2, 3, 4, 5, 6, 7, 8, 9]
productivity = [92, 88, 84, 78, 72, 65, 60, 52]
```

```python
plt.plot(months, study_hours, label = "study hours changed from January to June.")
plt.xlabel("Months")
plt.ylabel("Time")
plt.legend()
```

```python
plt.bar(study_hours, workout_hours)
plt.xlabel("study time")
plt.ylabel("workout time")
plt.show()
```

```python
plt.hist(marks, bins= 5)
plt.xlabel("marks")
plt.ylabel("Students")
plt.show()
```

```python
plt.scatter(screen_time, productivity)
plt.xlabel("Time")
plt.ylabel("Work done")
```

```python
plt.subplot(3,1,1)
plt.plot([1,2,3,4,5,6],study_hours)
plt.subplot(3,1,2)
plt.plot([1,2,3,4,5,6], workout_hours)
plt.subplot(3,1,3)
plt.plot([1,2,3,4,5,6], sleep_hours)
```
