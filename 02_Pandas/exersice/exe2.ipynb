```python
import pandas as pd
```

```python
tech = {
    "Laptop":65000,
    "Phone": 35000,
    "Tablet":24000,
    "Watch": 9000
}
acc = pd.Series(tech)
acc
```

```python
acc.max()
```

```python
acc.min()
```

```python
acc.mean()
```

```python
costly = acc[acc>20000]
costly
```

```python
highest = acc.idxmax()
lowest = acc.idxmin()
print(f"{highest} costs ₹{acc[highest]}")
print(f"{lowest} costs ₹{acc[lowest]}")
```

```python
salary = pd.Series({
    "Alice":50000,
    "Bob":75000,
    "Charlie":68000,
    "David":90000
})
salary
```

```python
highest = salary.idxmax()
lowest = salary.idxmin()
print(f"{highest} gets {salary[highest]} amount of salary and thats highest")
```

```python
print(f"{lowest} gets {salary[lowest]} amount of salary and thats least")
```

```python
students = pd.Series({
    "Ram":85,
    "Shyam":92,
    "Hari":76,
    "Ravi":88,
    "Sita":95,
    "Gita":67
})
students
```

```python
students.mean()
```

```python
hig_marks = students.max()
hig_marks
```

```python
hig_person = students.idxmax()
hig_person
```

```python
print(f"{hig_person} got highest marks : {hig_marks}")
```

```python
low_marks = students.min()
low_marks
```

```python
low_person = students.idxmin()
low_person
```

```python
print(f"{low_person} got highest marks : {low_marks}")
```
